Consider the following problem
There is a 5x5 grid where Mario lays on the bottom left square and bowser on the top right.
Every turn they both move simultaneously however only a preset of moves is valid.
For Mario he can only move up or to the right with equal probability, conversely bowser can only move down or to the left also with equal probability.
Should either reach the boundaries of the grid their next move will be the only possible one.
What is the probability that Mario and Bowser occupy the same square at any point in the game?

The following is my solution:
First off we quickly realize the max amount of turns that can occur is 8 forcing both into the opposing corners. However this is not relevant as only after 4 and exactly 4 moves is it possible for Mario and Bowser to share a square. After 4 moves Mario and Bowser can only be on 5 possible squares of which lie diagonally downwards across the grid as outlined by the image below. 

<img src="grid.png" width="150"/>

To calculate the probability of Mario reaching each of those squares we could do a quick count to measure the number of valid paths for each square.
We find the number of valid paths corresponding to the squares from left to right or up to down being 1, 4, 6, 4, 1
(Upon closer inspection we can note that this pattern is actually the equivelent of base 5 of the pascal triangle which makes sense as the number of valid paths grow additively with respect to past possible paths, an attribute that is shared with the fibonacci sequence and pascal triangle.)
Moving forth we must find the probabilities of Mario landing on each square which is as easy as dividng by the total number of possible paths: 16.
After that we square each probability (to take into account Bowsers respective chance of landing on the same square) and add them up resulting in this answer:  

$$\left(\frac{1}{16}\right)^2 + \left(\frac{4}{16}\right)^2 + \left(\frac{6}{16}\right)^2 + \left(\frac{4}{16}\right)^2 + \left(\frac{1}{16}\right)^2 = \frac{70}{256} = \frac{35}{128} \approx 27.3\%$$

The following code simulates the proposed problem and verifies my solution to be correct.

```python
import random


n = 670000
numhits = 0
for i in range (n):

    mariox = 1
    marioy = 1
    bowserx = 5
    bowsery = 5
        
    while mariox != 5 and marioy != 5 and bowserx != 1 and bowsery != 1:
 
        if mariox == 5 and marioy != 5:
            marioy += 1
        elif marioy ==5 and mariox != 5:
            mariox += 1
        else:
            if random.random() > .5:
                mariox += 1
            else:
                marioy += 1

        if bowserx == 1 and bowsery != 1:
            bowsery -= 1
        elif bowsery == 1 and bowserx != 1:
            bowserx -= 1
        else:
            if random.random() > .5:
                bowserx -= 1
            else:
                bowsery -= 1

        if mariox == bowserx and marioy == bowsery:
            numhits += 1
print(numhits/n)
```
